```python
import csv
import json

users = []

with open('glbids.csv', newline='') as f:
    reader = csv.reader(f, delimiter=';')
    for row in reader:
        _id = row[0]
        administrador = row[1]
        users.append(
            {
                "_Id": _id,
                "administrador": administrador
            }
        )

with open('users.json', 'w') as json_file:
    json.dump(users, json_file)
```
