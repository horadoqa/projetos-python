```python
def mostrar_perfil():
    print('{:-^40}'.format(''))
    print('{:-^40}'.format(' Perfil do Usuário '))
    print('{:-^40}'.format(''))
    print('Nome: {}'.format('João'))
    print('Idade: {}'.format(30))
    print('Profissão: {}'.format('Desenvolvedor'))
    print('Status: {}'.format('Ativo'))
    print('{:-^40}'.format(''))

mostrar_perfil()
```

Com parâmetros

```python
def banner():
    print('{:-^40}'.format(''))
    print('{:-^40}'.format(' Perfil do Usuário '))
    print('{:-^40}'.format(''))

def mostrar_perfil(nome, idade, profissao, status):
    banner()
    print(f'Nome: {nome}')
    print(f'Idade: {idade}')
    print(f'Profissão: {profissao}')
    print(f'Status: {status}')
    print('{:-^40}'.format(''))

mostrar_perfil('João', 30, 'Desenvolvedor', 'Ativo')
mostrar_perfil('Maria', 25, 'Designer', 'Inativo')
mostrar_perfil('Carlos', 40, 'Gerente', 'Ativo')
```
