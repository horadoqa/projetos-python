# Expressão Regular

## 1. URL

Validar uma URL: https://horadqa.com.br

```python
import re

def validar_url(url):
    padrao_url = r'^(https?://)?(www\.)?[a-zA-Z0-9-]+(\.[a-zA-Z]{2,})+(/[\w\.-]*)?$'

    if re.match(padrao_url, url):
        print(f"A URL '{url}' está correta!")
    else:
        print(f"A URL '{url}' não é válida.")

# Teste da função
url = input("Digite a URL para validar: ")

validar_url(url)
```

## 2. CPF

Exemplo: 52998224725

```python
import re

def validar_cpf(cpf):
    # Remove pontos e hífen
    cpf = re.sub(r'\D', '', cpf)

    # CPF precisa ter 11 dígitos
    if len(cpf) != 11:
        return False

    # Rejeita CPFs com todos os dígitos iguais
    if cpf == cpf[0] * 11:
        return False

    # Calcula o primeiro dígito verificador
    soma = sum(int(cpf[i]) * (10 - i) for i in range(9))
    resto = soma % 11
    digito1 = 0 if resto < 2 else 11 - resto

    # Calcula o segundo dígito verificador
    soma = sum(int(cpf[i]) * (11 - i) for i in range(10))
    resto = soma % 11
    digito2 = 0 if resto < 2 else 11 - resto

    # Verifica os dois dígitos
    return cpf[-2:] == f'{digito1}{digito2}'


cpf = input("Digite o CPF: ")

if validar_cpf(cpf):
    print(f"O CPF '{cpf}' é válido!")
else:
    print(f"O CPF '{cpf}' não é válido.")
```

## 3. Telefone

Validando um telefone brasileiro no formato (11) 99999-9999:

```python
import re

def validar_telefone(telefone):
    padrao = r'^\(\d{2}\) \d{5}-\d{4}$'

    if re.match(padrao, telefone):
        print("Telefone válido!")
    else:
        print("Telefone inválido.")


telefone = input("Digite o telefone: ")
validar_telefone(telefone)
```

## 4. CEP

Para o formato 12345-678:

```python
import re

def validar_cep(cep):
    padrao = r'^\d{5}-\d{3}$'

    if re.match(padrao, cep):
        print("CEP válido!")
    else:
        print("CEP inválido.")


cep = input("Digite o CEP: ")
validar_cep(cep)
```

## 5. E-mail

```python
import re

def validar_email(email):
    padrao = r'^[\w\.-]+@[\w\.-]+\.[a-zA-Z]{2,}$'

    if re.match(padrao, email):
        print("E-mail válido!")
    else:
        print("E-mail inválido.")


email = input("Digite o e-mail: ")
validar_email(email)
```

## 6. Data

Para o formato dd/mm/aaaa:

```python
import re

def validar_data(data):
    padrao = r'^(0[1-9]|[12]\d|3[01])/(0[1-9]|1[0-2])/\d{4}$'

    if re.match(padrao, data):
        print("Formato da data válido!")
    else:
        print("Data inválida.")


data = input("Digite a data: ")
validar_data(data)
```

## 7. Senha

Por exemplo, exigir:

- pelo menos 8 caracteres;
- uma letra maiúscula;
- uma letra minúscula;
- um número.

```python
import re

def validar_senha(senha):
    padrao = r'^(?=.*[A-Z])(?=.*[a-z])(?=.*\d).{8,}$'

    if re.match(padrao, senha):
        print("Senha válida!")
    else:
        print("Senha inválida.")


senha = input("Digite a senha: ")
validar_senha(senha)
```

## 8. CNPJ

Para validar inicialmente o formato 12.345.678/0001-90:

```python
import re

def validar_cnpj(cnpj):
    padrao = r'^\d{2}\.\d{3}\.\d{3}/\d{4}-\d{2}$'

    if re.match(padrao, cnpj):
        print("Formato do CNPJ válido!")
    else:
        print("Formato do CNPJ inválido.")


cnpj = input("Digite o CNPJ: ")
validar_cnpj(cnpj)
```
