```python
from rich.progress import Progress
import time

with Progress() as progress:
    task = progress.add_task("[cyan]Downloading...", total=100)
    while not progress.finished:
        progress.update(task, advance=1)
        time.sleep(0.30)
```

```python
from rich.progress import Progress
import time

TOTAL = 100
DELAY = 0.03

with Progress() as progress:
    task = progress.add_task("[cyan]Downloading...", total=TOTAL)

    while not progress.finished:
        progress.update(task, advance=1)
        time.sleep(DELAY)
```

```python
from rich.progress import (
    Progress,
    BarColumn,
    TextColumn,
    TaskProgressColumn,
    TimeRemainingColumn,
)
import time

with Progress(
    TextColumn("[cyan]{task.description}"),
    BarColumn(),
    TaskProgressColumn(),
    TimeRemainingColumn(),
) as progress:

    task = progress.add_task("Downloading...", total=100)

    while not progress.finished:
        time.sleep(0.03)
        progress.update(task, advance=1)
```

```python
import requests
from rich.progress import (
    Progress,
    BarColumn,
    DownloadColumn,
    TransferSpeedColumn,
    TimeRemainingColumn,
)
from rich.console import Console

URL = "https://serverest.dev/usuarios"
ARQUIVO = "usuarios.json"

console = Console()

try:
    with requests.get(URL, stream=True, timeout=30) as response:
        response.raise_for_status()

        total = int(response.headers.get("content-length", 0))

        with Progress(
            "[progress.description]{task.description}",
            BarColumn(),
            "[progress.percentage]{task.percentage:>3.0f}%",
            DownloadColumn(),
            TransferSpeedColumn(),
            TimeRemainingColumn(),
        ) as progress:

            task = progress.add_task(
                "Downloading",
                total=total if total else None,
            )

            with open(ARQUIVO, "wb") as arquivo:
                for chunk in response.iter_content(chunk_size=8192):
                    if chunk:
                        arquivo.write(chunk)
                        progress.update(task, advance=len(chunk))

    console.print(f"[green]✓ Download concluído:[/green] {ARQUIVO}")

except requests.RequestException as e:
    console.print(f"[red]✗ Erro no download:[/red] {e}")

except OSError as e:
    console.print(f"[red]✗ Erro ao salvar o arquivo:[/red] {e}")
```
